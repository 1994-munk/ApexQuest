# ApexQuest Games - Player Analysis Notebook

Interactive exploration of player statistics and performance metrics.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
from pathlib import Path

# Add src to path
sys.path.insert(0, str(Path.cwd().parent / 'src'))

from data_loader import load_all_data, clean_data
from analysis import PlayerAnalyzer
from visualization import PlayerVisualizer

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
%matplotlib inline
```

## Load and Prepare Data

```python
# Load player data
df = load_all_data('../data')
df = clean_data(df)
print(f"Loaded {len(df)} players")
df.head()
```

## Player Statistics

```python
analyzer = PlayerAnalyzer(df)
analyzer.get_player_stats_summary()
```

## Performance Analysis

```python
# Calculate KDA ratio
kda = analyzer.calculate_kda_ratio()
print("Top 10 Players by KDA Ratio:")
print(kda.head(10))
```

```python
# Visualize KDA distribution
plt.figure(figsize=(10, 6))
sns.histplot(data=analyzer.df, x='kda_ratio', bins=20, kde=True)
plt.title('KDA Ratio Distribution')
plt.xlabel('KDA Ratio')
plt.ylabel('Number of Players')
plt.tight_layout()
plt.savefig('../outputs/kda_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Player Engagement

```python
engagement = analyzer.get_engagement_stats()
for key, value in engagement.items():
    print(f"{key}: {value}")
```

```python
# Playtime vs Kills
if 'playtime_hours' in analyzer.df.columns and 'kills' in analyzer.df.columns:
    plt.figure(figsize=(10, 6))
    plt.scatter(analyzer.df['playtime_hours'], analyzer.df['kills'], alpha=0.6)
    plt.xlabel('Playtime (hours)')
    plt.ylabel('Kills')
    plt.title('Playtime vs Kills')
    plt.tight_layout()
    plt.savefig('../outputs/playtime_vs_kills.png', dpi=300, bbox_inches='tight')
    plt.show()
```
